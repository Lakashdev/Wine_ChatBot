```python
import pandas as pd

# Suppose df["Image_URL"] looks like the example
def pick_best_image(cell):
    if pd.isna(cell):
        return None
    urls = str(cell).split("|")
    urls = [u.strip() for u in urls if u.strip()]

    if not urls:
        return None

    # 1) Prefer jpg/jpeg over webp/png
    for u in urls:
        if u.lower().endswith((".jpg", ".jpeg")):
            return u
    for u in urls:
        if u.lower().endswith(".png"):
            return u
    for u in urls:
        if u.lower().endswith(".webp"):
            return u

    # fallback: first one
    return urls[0]

# Cleaned column
df["Image_URL_clean"] = df["Image_URL"].apply(pick_best_image)
```

```python
print(df[['Image_URL', 'Image_URL_clean']].head(20))
```

```python
# Rows that had multiple links originally
multi_links = df['Image_URL'].astype(str).str.contains('|', regex=False).sum()
print("Rows with multiple URLs originally:", multi_links)

# Rows still with multiple links after cleaning (should be 0)
multi_cleaned = df['Image_URL_clean'].astype(str).str.contains('|', regex=False).sum()
print("Rows with multiple URLs after cleaning:", multi_cleaned)
```
